```python
import numpy as np
import pandas as pd
import matplotlib as mlt
```

## Cleaning Customers File

```python
Customers = pd.read_csv('Customers.csv')
Customers.head(10)
```

## Cleaning the Inconsistance Data

```python
# Cleaning Gender Columns
Customers['Gender'].unique()
```

```python
Customers['Gender']=Customers['Gender'].str.title().str.strip()
Customers['Gender'].unique()
```

```python
Customers['Gender']=Customers['Gender'].replace({'M':'Male','F':'Female','Fem':'Female'})
Customers['Gender'].unique()
```

```python
#Cleaning Country Column
Customers['Country'].unique()
```

```python
Customers['Country']=Customers['Country'].str.title().str.strip()
Customers['Country'].unique()
```

```python
Customers['Country']=Customers['Country'].replace({'Usa':'USA','Bharat':'India','Ind':'India','Uk':'UK','Uae':'UAE'})
Customers['Country'].unique()
```

```python
# Changing Column Name
Customers.rename(columns={'STATE': 'State'}, inplace=True)
```

```python
#Cleaning The State Columns
Customers['State'].unique()
```

```python
Customers['State'] = Customers['State'].str.title().str.strip()
Customers['State'].unique()
```

```python
Customers['State']=Customers['State'].replace({'Dl':'Delhi','Tn':'Tamil Nadu','Tamilnadu':'Tamil Nadu','Maharastra':'Maharashtra',
                                               'Mh':'Maharashtra','Ka':'Karnataka'})
Customers['State'].unique()
```

## Handling missing values 

```python
# As 16.9% gender data is null so we can't drop the null 
Customers['Gender'].value_counts()
```

```python
#Filling the null values with most repeated values
most_gender = Customers['Gender'].mode()[0]
Customers['Gender']=Customers['Gender'].fillna(most_gender)
```

```python
Customers.isnull().sum()/len(Customers)*100
```

```python
# Cleaning the Age Columns
Customers['Age'].unique()
```

```python
# Converting strings into numbers
from word2number import w2n
Customers['Age']=Customers['Age'].apply(lambda x : w2n.word_to_num(x) if isinstance(x,str) else x)
```

```python
Customers['Age'].unique()
```

## Changing the Data Types

```python
#Age Columns object to int
Customers['Age'] = Customers['Age'].astype('int64')
```

```python
# Date Columns Object to Date
Customers['Signup_Date']=pd.to_datetime(Customers['Signup_Date'])
```

```python
Customers.info()
```

```python
Customers.head(10)
```

## Summary

* Cleaned the incorrect formats into single formats using replace
* As there is 16.9%  null gender so we can't drop it so in that case i fill null values with most repeated values in gender
* Some Ages are into word format so by using w2n library conver the words into numbers
* In the end conver age into int64 and shipping date into datetime data type

# Cleaning Product Files

```python
# importing product files

Products = pd.read_csv('products.csv')
Products.head(10)
```

## Cleaning the Inconsistance Data

```python
#Cleaning Product_Name Columns
Products['Product_Name'].unique()
```

```python
Products['Product_Name']=Products['Product_Name'].str.title().str.strip()
Products['Product_Name'].unique()
```

```python
#Cleaning Category Columns
Products['Category'].unique()
```

```python
Products['Category']=Products['Category'].str.title().str.strip()
Products['Category'].unique()
```

## Cleaning numerical Data Columns

```python
# Correcting the negative values of Price Columns
Products['Price']=Products['Price'].abs()
Products['Price'].describe()
```

```python
Products[['Price', 'Cost']].describe()
```

```python
Products[Products['Cost'] < Products['Price']]
```

```python
#Rounding The Integer values
Products[['Cost', 'Price']]=Products[['Cost', 'Price']].round(2)
```

#### There are still Some columns which contains Price value greater than Cost Value

```python
## Swaping the values of price with cost where the price is greater than cost 
mask = Products['Cost'] < Products['Price']

Products.loc[mask, ['Price','Cost']] = Products.loc[mask, ['Cost','Price']].values
```

```python
Products[Products['Cost'] < Products['Price']]
```

```python
Products[['Price', 'Cost']].describe()
```

```python
Products.info()
```

```python
Products.head(10)
```

### Summary

* Cleaned the insconsitence data of product name and category by formating the data 
* Convert Negative values to positive by using abs
* Swap all the values of price with cost where price is greated than cost using map

```python
#Re-Checking the condition
Order_items[Order_items['Quantity']*Order_items['Price_per_Unit']!= Order_items['Total_Amount']]
```
